Data Wrangling : dealing with missing value, reshape your data

```python
#assign data

data={'Name': ['Yamini','Pratik','Pratiksha','Avinash'],
      'Age' : [20,22,24,25],
      'Gender' : ['F','M','F','M'],
      'Marks' : [90,95,'NAN',99]}
df=pd.DataFrame(data)
df
```

```python
# Dealing with missing value
c=avg=0
for i in df ["Marks"]:
  if str(i).isnumeric():
    c += 1
    avg += i
avg /= c
df = df.replace (to_replace = "NAN", value = avg)
df
```
